# Matplotlib

```python
import matplotlib
import matplotlib.pyplot as plt

print(matplotlib.__version__)
```

```python
import numpy as np
import matplotlib.pyplot as plt
x = np.arange(0, 10)
y = x**2
plt.plot(x, y)
plt.title('This is title')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.show()
```

```python
x = [ i**2 for i in range(1,9)]

plt.plot(x)
plt.plot(x, color='r')
plt.plot(x, 'or')
plt.plot(x, '-k')
plt.plot(x, '^b')
plt.plot(x, '--oy')

plt.show()
```

```python
x = np.arange(0, 10)
y = x**2
plt.plot(x, y)
plt.title('This is title')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.show()
```

```python
x = np.linspace(0, 10, 100)
y = np.sin(x)
z = np.cos(x)
plt.plot(x, y)
plt.plot(x, z)
plt.show()
```

```python
x = np.linspace(0, 10, 30)
y = np.sin(x)
z = np.cos(x)
fig, ax = plt.subplots(2, 2)
fig.set_size_inches(10, 8)
ax[0][0].plot(x, y, 'r-')
ax[0][1].plot(x, z, 'g-')
ax[1][0].plot(x, z, 'b-')
plt.show()
```

```python
a = np.arange(0,9)
b = [1,4,6,9,5,2,7,3,8]

plt.figure(figsize=(10,6))
plt.plot(a,b,color='r', linestyle='--', marker='o', markerfacecolor='b', markersize=10)

plt.xlim([3,7])

plt.show()
```

```python
a = np.arange(0,10)
b = np.array([9,4,6,9,5,2,7,1,3,2])

plt.figure(figsize=(10,6))
plt.scatter(x=a, y=b, s=100, c=a, marker='^')
plt.colorbar()
plt.show()
```

```python
a = np.random.normal(loc=0, scale=1, size=1000)
b = np.random.normal(loc=5, scale=0.5, size=1000)
c = np.random.normal(loc=10, scale=2, size=1000)

plt.figure(figsize=(10,6))
plt.plot(a, label='a')
plt.plot(b, label='b')
plt.plot(c, label='c')

plt.legend()
plt.show()
```

```python
a = [3,2,1,4,3,1,5,2,12,1]

import pandas as pd
df = pd.DataFrame({'a':a})
df.mean()
df.median()
```

```python
import statistics

mean_val = statistics.mean(a)
median_val = statistics.median(a)

print(f"평균값: {mean_val}")
print(f"중앙값: {median_val}")
```

```python
a = np.random.normal(loc=0, scale=1, size=1000)
b = np.random.normal(loc=5, scale=0.5, size=1000)
c = np.random.normal(loc=10, scale=2, size=1000)

plt.figure(figsize=(10,6))
plt.boxplot((a,b,c))

plt.legend()
plt.show()
```

```python
name = ['Jane','Alice','Tom','John']
score = [85,100,75,60]

plt.figure(figsize=(10,6))
plt.bar(name, score)
a = np.arange(len(name))
sc1 = plt.subplot()
fig, sc1 = plt.subplots()
sc1.bar(a,score,color='green')
plt.show()
```

```python
x = np.linspace(0,10,100)
y = np.sin(x)

plt.style.use('ggplot')

plt.plot(x,y,'b-', linewidth=3)
plt.xticks(np.arange(0,10,2))
plt.yticks(np.arange(-1.0,1.1))

plt.show()
```
