# 넘파이(numpy)

[NumPy](https://numpy.org/)
- ndarray 객체, 다차원 배열
- **넘파이**는 메모리를 효율적으로 사용하기 위해 하나의 타입만 들어갈 수 있음
  - cf. **리스트**는 여러가지 타입이 들어갈 수 있음(숫자와 문자)

- Numpy란
  - Numerical Python의 약자.
  - 다차원 배열을 효율적으로 처리하고 쉽게 사용할 수 있도록하는 파이썬 라이브러리.
  - 데이터 구조 외에도 수치 계산을 위해 효율적으로 구현된 기능을 제공.
  - 파이썬으로 진행하는 데이터 분석과 인공지능 학습에 있어 가장 필수적인 라이브러리.
  - 파이썬의 List에 비해 빠르고, 메모리를 효율적으로 사용.
  - 반복문 없이 배열의 처리를 지원하므로 빠르고 편리함.

```python
import numpy as np
print(np.__version__)
```

- ndarray 객체
  - N-dimensional Array의 약자.
  - ndarray객체는 넘파이에서 다차원 배열을 의미.
  - 모든 원소가 같은 자료형이어야 한다.
  - 한 번 생성된 배열은 원소의 개수 변결 불가.

```python
d1 = [1,2,'3',4,5]
print(type(d1),type(d1[1]),type(d1[2]))
```

```python
a1 = np.array(d1)
print(a1)
print(type(a1),a1.shape,a1.dtype,a1.itemsize)
```

```python
d2 = [1,2,3,4,5]
a2 = np.array(d2)
print(type(a2),a2.shape,a2.dtype,a2.itemsize)
```

- 형변환 처리

```python
d8 = list(range(1,13))
d8
a8_1 = np.array(d8)
print(a8_1)
a8_2 = np.array(d8).reshape(2,6)
print(a8_2)
a8_3 = np.array(d8).reshape(3,2,2)
print(a8_3)

print(a8_1.shape,a8_2.shape,a8_3.shape)
print(a8_1.ndim,a8_2.ndim,a8_3.ndim)
```

```python
# 3 8 11 출력
print(a8_3[0][1][0],a8_3[1][1][1],a8_3[2][1][0])
```

- a9 = a9.astype(np.string_) #<String
- 이 오류는 np.string_ 데이터 타입이 NumPy 2.0 릴리스에서 제거되었기 때문에 발생.
- 이제 대신 np.bytes_를 사용해야 함.

```python
a9 = np.arange(1,7).reshape(2,3)
print(a9,a9.dtype)
a9 = a9.astype(np.float64)
print(a9,a9.dtype)
a9 = a9.astype('int32')
print(a9,a9.dtype)
a9 = a9.astype('float64')
print(a9,a9.dtype)
a9 = a9.astype('str') #<U32
print(a9,a9.dtype,a9.itemsize)

# a9 = a9.astype(np.string_) #<String
a9 = a9.astype(np.bytes_) #<String
print(a9,a9.dtype,a9.itemsize)

a9 = a9.astype(np.str_) #<U32
print(a9,a9.dtype,a9.itemsize)
```

- 함수들
  - np.arrange()
  - np.ones()
  - np.linspace()
  - np.zeros()
  - np.full()

```python
a1 = np.arange(1,26).reshape(5,5)
print(a1)

print(a1[2][2], a1[2,2])
print(a1[1:-1])
print(a1[1:-1,1:-1])
print(a1[3:,:3])
print(a1[:1,-1:])

a2 = np.ones(9)
print(a2)

a2 = np.ones(9).reshape(3,3)
print(a2)

a3 = np.zeros(9).reshape(3,3)
print(a3)

a3 = np.eye(5)
print(a3)

# 주어진 구간을 균일한 간격으로 나눈다..
# np.linspace(시작, 끝, 개수).
# 시작과 끝을 포함해서 지정된 개수만큼 균일한 간격으로 숫자를 채움.
a4 = np.linspace(0, 4, 5)
print(a4)

# 주어진 형태의 배열을 만들고, 모두 주어진 값으로 채움.
a5 = np.full((2,3),0.5)
print(a5)

# ones_like(), zeros_like() 함수
# 인수로 지정된 배열 객체의 형태(shape)와 같은 형태의 배열을 생성하고 초기화.
a6 = np.array([[1, 2], [3, 4]])
print(a6.shape)
a7 = np.zeros_like(a6)
print(a7, a7.shape)
a8 = np.ones_like(a6)
print(a8, a8.shape)
```

- 배열(Array)의 축(Axis)

```python
import numpy as np

x = np.array([1, 2, 3, 4, 5])
y = np.array([[1, 2, 3, 4, 5]])
z = np.array([[1], [2], [3], [4], [5]])
print(x.shape, y.shape, z.shape)
```

- 전치행렬(transpose matrix) :
  - 행렬의 행과 열을 바꾸기, 행렬의 축을 바꾸는 방법

```python
a4 = np.arange(1,7).reshape(2,3)
print(a4)

a5 = np.transpose(a4)
print(a5)

a6 = a5.T
print(a6)
```

- 연산

```python
a1 = np.array([[1,2],[3,4]])
a2 = np.array([10,20])
a3 = np.array([3,4])

print(a2 + 1)
print(a2 - 30)
print(a2 * 10)
print(a2 / 2)
print(100 / a2)

print(a1 + 1)
print(a1 - 30)
print(a1 * 10)
print(a1 / 2)
print(100 / a1)

print(a2 + a3)
print(a1 + a2)

print(a1 * a2)
```

- 행렬곱 : matmul( , )

```python
a3 = np.array([[1,2],[3,4]])
a4 = np.array([[2,3],[4,5]])

print(a3 + a4)
print(a3 * a4)
print(np.matmul(a3,a4)) # 행렬 곱

a5 = np.arange(1,101)
print(a5.max(),a5.min(),a5.mean(),a5.sum())

a5 = np.arange(1,101).reshape(10,10)
print(a5.max(),a5.min(),a5.mean(),a5.sum())
```
