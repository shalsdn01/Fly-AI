- 가변길이 : *

```python
def Sum(*a):
    ans = 0
    for i in a:
        ans += i
    return ans

print(Sum(20,10))   # 30
print(Sum(20,10,5))   # 35
print(Sum(20,10,5,3))   # 38
print(Sum(20,10,5,3,2))   # 40

b = [11,22,33,44]
print(Sum(*b))
```

- 가변길이 : ** (dict)

```python
a = {'사과':'apple',
     '포도':'grape',
     '바나나':'banana',
     '컵':'cup',
     '물':'water'}

def pDict(**d):
    for i in d.items(): print(i)

pDict(**a)
pDict(**{'컵':'cup','물':'water'})
```

- 고정길이, 가변길이

```python
def pSum1(s,*a):
    ans = 0
    for i in a:
        ans += i
    return '{}: {}'.format(s,ans)


print(pSum1('덧셈',20,10)) # 덧셈 : 30
print(pSum1('덧셈',20,10,5)) # 덧셈 : 35
print(pSum1('덧셈',20,10,5,3)) # 덧셈 : 38

def pSum2(*a,s):
    ans = 0
    for i in a:
        ans += i
    return '{}: {}'.format(s,ans)

print(pSum2(20,10,s='덧셈')) # 덧셈 : 30
```

- 선언적 함수, 람다식(Lambda)

```python
def Sum(i,j): return i+j # 선언적 함수
def Sub(i,j): return i-j
def Mul(i,j): return i*j
def Div(i,j): return i/j

cList = [Sum,Sub,Mul,Div]
for c in cList:
    print(c(200,100))

cList2 = []
cList2.append(Mul)
cList2.append(Sub)
cList2.append(Sum)

for c in cList2:
    print(c(2000,1000))

cList2.remove(Sub)
for c in cList2:
    print(c(2000,1000))
```

```python
# 람다식 : x,y -> x+y

cList = [Sum,Sub,Mul,Div,lambda x,y:x%y]
for c in cList:
    print(c(20,10))

Nmj = lambda x,y:x%y
print(Nmj(40,6))
```

- 자주 사용하는 함수들

```python
import re
# replace, join, split, map, zip, filter, reduce

a = 'Pithon'
# a[1] = 'y' # 변경 불가
a = a.replace('i', 'y')
a

b = 'Welcome to Korea Education'
print(b.replace('co', 'xt'))
print(re.sub('co', 'xt',b))
print(b)

c = ['k','o','r','e','a']
print(''.join(c))
print('/'.join(c))

d = 'Java,SQL,OS,Python'
print(d.split(','))

# re.split(pattern, string) : 정규표현식에도 존재

e = ['10','20','30','40']
print(list(map(int,e)))

f1 = ['콜라','사이다','우유','활명수']
f2 = [1000,900,600,700]
print(dict(zip(f1,f2)))
print(list(zip(f1,f2)))

g1 = [8,3,2,5,7,9,11,15,4,12]
#g1 -> g2 5보다 크고, 홀수인 값 이전
g2 = [ i for i in g1 if (i>5) & (i%2==1) ]

def func1(x):
    return (x>5) & (x%2==1)

g3 = list(filter(func1,g1))
g3

h1 = [8,3,2,5,7,9,11,15,4,12]
# h1 -> h2 : 5보다 크고, 홀수인 값만 제곱 이전
h2 = [ i**2 for i in g1 if (i>5) & (i%2==1) ]
h2

def func2(x):
    return x**2

h3 = list(map(func2,filter(func1,h1)))
h3

j1 = [8,3,2,5,7,9,11,15,4,12]

def func11(x): return x>5
def func12(x): return x%2==1
def func13(x): return x**2

j2 = list(map(func13,filter(func12,filter(func11,j1))))
print(j2)

from functools import reduce

def func21(x,y):
    return x+(y**2)

k1 = [1,2,3,4,5]
k2 = reduce(func21,k1)
k2 # 55
```
