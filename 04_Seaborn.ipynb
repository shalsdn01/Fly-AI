# Seaborn

```python
import seaborn as sns
print(sns.__version__)
```

- seaborn 함수들은 크게 'figure-level' 또는 'axes-level'함수로 나뉘어져 있음

```python
filepath = 'https://raw.githubusercontent.com/mwaskom/seaborn-data/refs/heads/master/tips.csv'
import pandas as pd
tips = pd.read_csv(filepath)

import matplotlib.pyplot as plt

sns.countplot(data=tips, x='day', hue='smoker')
plt.show()
```

```python
print(tips['day'].value_counts())
```

```python
sns.barplot(data = tips, x='sex', y='tip', hue='sex')
plt.show()
```

```python
tips.groupby('sex')['tip'].mean()
```

```python
import numpy as np
sns.barplot(data = tips, x='sex', y='tip', hue='sex', estimator=np.median)
plt.show()
```

```python
sns.histplot(x='total_bill', data=tips, bins=14, kde=True, color='green')
plt.show()
```

```python
sns.displot(data=tips, x='total_bill', kde=True, color='green')
plt.show()
```

```python
sns.displot(data=tips, x='total_bill', kde=True, col='time')
plt.show()
```

```python
sns.boxplot(x='smoker', y='total_bill', data=tips, hue='smoker')
plt.show()
```

```python
sns.violinplot(x='smoker', y='total_bill', data=tips, hue='sex')
```

```python
df = tips.select_dtypes(['number'])
x = df.corr()
sns.heatmap(x, annot=True, cmap='RdYlBu')
plt.show()
```

```python
df = sns.load_dataset('titanic')
sns.heatmap(df.isnull(), yticklabels=False,
 cbar=False, cmap='viridis')
```

```python
g = sns.FacetGrid(tips, col="time", row="sex")
g.map(sns.scatterplot, "total_bill", "tip")
plt.show()
```

```python
g = sns.FacetGrid(tips, col="time", row="sex")
g.map_dataframe(sns.histplot, x="total_bill")
plt.show()
```
